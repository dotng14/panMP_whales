# panMP on LeiData 014845 at 10 kHz

`panMP` on the first 60 seconds of `SOCAL_H_76_230427_014845.x_10kHz.wav` (10 kHz, 37.5 minutes total). A pairwise profile of all 22.5 million samples is not practical.

Motif lengths are the `panMP` defaults: 32, 96, …, 480 samples (3.2–48 ms at 10 kHz), step 64, through 512. Distances come from `mpx_fast`. High surface score means a repeated motif.

```python
from pathlib import Path

from IPython.display import Image, Markdown, display

FIG = Path.home() / "whale_processing/result/LeiData/matrix_profile/014845_10kHz"
missing = [name for name in (
    "pan_surface.png",
    "pan_surface_3d.png",
    "matrix_profiles.png",
    "best_motif_pair.png",
) if not (FIG / name).is_file()]
if missing:
    raise FileNotFoundError(f"missing under {FIG}: {missing}")
print(FIG)
```
